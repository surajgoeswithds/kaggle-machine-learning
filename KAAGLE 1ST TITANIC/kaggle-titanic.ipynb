```python
import pandas as pd 
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
```

```python
load_path ='/kaggle/input/competitions/titanic/train.csv'
hold_titanic_data = pd.read_csv(load_path)
```

```python
hold_titanic_data.head()
hold_titanic_data.columns
hold_titanic_data.describe()
hold_titanic_data['Age']=hold_titanic_data['Age'].fillna(hold_titanic_data['Age'].median())
hold_titanic_data.head()
```

```python
hold_titanic_data.describe()
```

```python
hold_titanic_data['Sex'] = hold_titanic_data['Sex'].map({"male" : 0, "female" : 1})
hold_titanic_data.describe()
```

```python
y = hold_titanic_data.Survived
```

```python
features = ['Pclass' , 'SibSp' , 'Parch' , 'Fare' , 'Sex' , 'Age']
X = hold_titanic_data[features]
```

```python
train_X , val_X , train_y , val_y = train_test_split(X , y ,random_state=1)
titanic_model = RandomForestClassifier(random_state=1)
titanic_model.fit(train_X , train_y)
```

```python
titanic_predict = titanic_model.predict(val_X)
mae = mean_absolute_error(titanic_predict , val_y)
print(mae)
#MAE is 0.26 — this makes sense for a classification problem. It means model is wrong about 26% of the time on survival predictions.
```

```python
titanic_final_model = RandomForestClassifier(random_state = 1)
titanic_final_model.fit( X , y)
```

```python
load_path2 = '/kaggle/input/competitions/titanic/test.csv'
hold_titanic_data2 = pd.read_csv(load_path2)
```

```python
hold_titanic_data2['Age']=hold_titanic_data2['Age'].fillna(hold_titanic_data2['Age'].median())
hold_titanic_data2['Sex'] = hold_titanic_data2['Sex'].map({"male" : 0, "female" : 1})
```

```python
Z = hold_titanic_data2[features]
```

```python
predictions = titanic_final_model.predict(Z)
```

```python
output = pd.DataFrame({'PassengerId': hold_titanic_data2.PassengerId, 'Survived': predictions})
output.to_csv('submission.csv', index=False)
```
